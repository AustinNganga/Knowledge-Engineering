**Step 1: Identify the Task**

The task is to help a car decide what to do based on the traffic light color.

**Goal:** Make safe driving decisions with minimal risk.

**Step 2: Assemble Relevant Knowledge**

The system needs to know:
*   The color of the traffic light
*   The correct action to take for each light color

**Step 3: Define the Vocabulary**
*   Predicates (represented in Python as keys and rules): Light(color), CanGo(color)
*   Constants: Red, Yellow, Green

**Step 4: Encode General Knowledge about the Domain** `[3 marks]`

```
It's crucial for knowledge engineers to explicitly declare the vocabulary here. If you used APIs/web scraping tools to assemble  knowledge for the practical exercise, check step 2.
```

````python
rules = {
    "Red": "Stop — it’s unsafe to move.",
    "Yellow": "Slow down — prepare to stop.",
    "Green": "Go — it’s safe to proceed.",
}
````

**Step 5: Add New Facts to the Knowledge Base** `[4 marks]`

````python
def show_kb():
    print("\n--- Current Knowledge Base ---")
    for light, action in rules.items():
        print(f"{light:10} ➜ {action}")
    print("-------------------------------")

def add_new_fact(color, action):
    """Add a new fact directly to the KB (- without user input)."""
    rules[color.capitalize()] = action
    print(f"Added fact: '{color.capitalize()}' ➜ '{action}'")

# Example of adding facts
add_new_fact("Flashing", "Stop and check intersection")
add_new_fact("Blinking", "Proceed slowly with caution")

def add_fact():
    """Add new knowledge (facts) to the KB."""
    new_light = input("Enter new traffic light color: ").capitalize()
    new_action = input(f"What should the agent do when light is {new_light}? ")
    rules[new_light] = new_action
    print(f" New fact added: '{new_light}' -> '{new_action}'")

add_fact()

# View updated KB
show_kb()
````

**Steps 6 & 7: Encode Specific Problem Instance (Facts) & Pose Queries and Get Answers** `[3 marks]`



````python
# Step 6: The current observation from sensors/camera
current_light = input("Enter the current traffic light color: ").capitalize()

# Step 7: The agent queries the knowledge base (rules dictionary)
def traffic_advice(light):
    action = rules.get(light, "Unknown signal — proceed with caution or stop.")
    return action

# Output the result
print("\n=== Traffic Light Decision System ===")
print(f"Observed Light: {current_light}")
print("Agent Decision:", traffic_advice(current_light))
````

**Step 8: Querying the knowledge base**

````python
#ask questions to get actions based on known facts (in this case; lights)
print("Agent Decision for 'Flashing':", traffic_advice("Flashing"))
print("Agent Decision for 'Red':", traffic_advice("Red"))
print("Agent Decision for 'Teal':", traffic_advice("Teal"))
````

**Step 9: Debugging the Knowledge Base** `[2 marks]`

````python
# You can test new or faulty signals to see how the agent reacts.
test_cases = ["Red", "Yellow", "Green", "Blue", "Teal", "None", "Purple"]
print("\n--- Debugging the Knowledge Base ---")
for color in test_cases:
    print(f"Light: {color:6} -> Action: {traffic_advice(color)}")
````
